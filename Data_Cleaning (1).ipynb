# **Handle Missing Quantity**

**Create a copy**

```python
clean_df = df.copy()
```

**Fill missing (Price Per Unit)**

```python
mask = (
    clean_df["Price Per Unit"].isnull() &
    clean_df["Quantity"].notnull() &
    clean_df["Total Spent"].notnull()
)

clean_df.loc[mask, "Price Per Unit"] = (
    clean_df.loc[mask, "Total Spent"] / clean_df.loc[mask, "Quantity"]
)
```

```python
clean_df["Price Per Unit"].isnull().sum()
```

**Fill missing (Total Spent)**

```python
mask = (
    clean_df["Total Spent"].isnull() &
    clean_df["Price Per Unit"].notnull() &
    clean_df["Quantity"].notnull()
)

clean_df.loc[mask, "Total Spent"] = (
    clean_df.loc[mask, "Price Per Unit"] * clean_df.loc[mask, "Quantity"]
)
```

```python
clean_df["Total Spent"].isnull().sum()
```

**Fill missing (Quantity)**

```python
mask = (
    clean_df["Quantity"].isnull() &
    clean_df["Price Per Unit"].notnull() &
    clean_df["Total Spent"].notnull()
)

clean_df.loc[mask, "Quantity"] = (
    clean_df.loc[mask, "Total Spent"] / clean_df.loc[mask, "Price Per Unit"]
)
```

```python
clean_df["Quantity"].isnull().sum()
```

```python
print("Missing Price:", clean_df["Price Per Unit"].isnull().sum())
print("Missing Quantity:", clean_df["Quantity"].isnull().sum())
print("Missing Total:", clean_df["Total Spent"].isnull().sum())
```

```python
clean_df[
    clean_df["Quantity"].isnull() &
    clean_df["Total Spent"].isnull()
].head(10)
```

```python
clean_df[
    clean_df["Quantity"].isnull() &
    clean_df["Total Spent"].isnull()
]["Category"].value_counts()
```

```python
clean_df[
    clean_df["Quantity"].isnull() &
    clean_df["Total Spent"].isnull()
]["Price Per Unit"].describe()
```

**Check whether Quantity can be reasonably imputed**

```python
clean_df.groupby("Category")["Quantity"].agg(["count", "median", "mean"])
```

**Check the actual quantity distribution**

```python
clean_df["Quantity"].value_counts().sort_index()
```

**One more important check for (Item)**

```python
clean_df[
    clean_df["Item"].isnull()
][["Category", "Price Per Unit"]].head(20)
```

```python
clean_df.groupby(
    ["Category", "Price Per Unit"]
)["Item"].nunique()
```

```python
clean_df["Quantity"] = clean_df["Quantity"].fillna(
    clean_df.groupby("Category")["Quantity"].transform("median")
)

clean_df["Quantity"] = clean_df["Quantity"].astype(int)
```

```python
clean_df["Total Spent"] = clean_df["Total Spent"].fillna(
    clean_df["Price Per Unit"] * clean_df["Quantity"]
)
```

```python
clean_df.isnull().sum()
```

```python
item_lookup = (
    clean_df.dropna(subset=["Item"])
    .drop_duplicates(["Category", "Price Per Unit"])
    .set_index(["Category", "Price Per Unit"])["Item"]
)
```

```python
mask = clean_df["Item"].isnull()

clean_df.loc[mask, "Item"] = [
    item_lookup.get((category, price))
    for category, price in zip(
        clean_df.loc[mask, "Category"],
        clean_df.loc[mask, "Price Per Unit"]
    )
]
```

```python
clean_df["Item"].isnull().sum()
```

```python
clean_df["Discount Applied"].value_counts(dropna=False)
```

**Investigate whether Discount Applied can be derived**

```python
clean_df["Calculated Total"] = (
    clean_df["Price Per Unit"] * clean_df["Quantity"]
)
```

```python
clean_df.groupby("Discount Applied")[
    ["Price Per Unit", "Quantity", "Total Spent", "Calculated Total"]
].mean()
```

```python
clean_df["Discount Difference"] = (
    clean_df["Calculated Total"] - clean_df["Total Spent"]
)
```

```python
clean_df.groupby("Discount Applied")["Discount Difference"].describe()
```

**Handle missing Discount Applied**

```python
clean_df["Discount Applied"] = clean_df["Discount Applied"].fillna("Unknown")
```

```python
clean_df["Discount Applied"].value_counts()
```

**Remove our temporary analysis columns**

```python
clean_df.drop(
    columns=["Calculated Total", "Discount Difference"],
    inplace=True
)
```
